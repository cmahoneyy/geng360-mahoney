```python
import numpy as np

vin = 1
f = np.array([1, 100, 720, 5000]) # Hz
r = 4.7e3 # k ohm to ohm
c = 47e-9 # uF to F

zc = 1 / (2 * np.pi * f * c) # in ohms

z = np.sqrt(r**2 + zc**2) # input signal of 1 V AC

vout = vin * (zc / z) # output voltage across the capacitor

print(vout) # print output voltages

```
